This assignment implements a Retrieval-Augmented Generation (RAG) system using LangChain to answer questions based on TED Talk transcripts. The system combines semantic search with transformer models to provide accurate, context-aware responses.

## 1. Import Required Libraries

```python
# Core libraries
from datasets import load_dataset
from langchain.schema import Document
from langchain.text_splitter import RecursiveCharacterTextSplitter
from langchain.embeddings import HuggingFaceEmbeddings
from langchain.vectorstores import FAISS
from langchain.llms import HuggingFacePipeline
from langchain.chains import RetrievalQA
from transformers import pipeline
import warnings
warnings.filterwarnings('ignore')
```

## 2. Document Loading & Preprocessing

Loading TED Talks dataset and converting to LangChain Document format.

```python
dataset = load_dataset("ted_talks_iwslt", "nl_en_2016", trust_remote_code=True)
print(f"Dataset loaded with {len(dataset)} examples")
```

```python
# Examine dataset structure
print("Dataset structure:")
print(dataset)
print("\nSample item:")
print(dataset[0])
```

```python
# Convert dataset to LangChain Documents
print("Converting to LangChain Documents...")
documents = []

for i, item in enumerate(dataset):
    # Extract English translation
    if item["translation"] and "en" in item["translation"]:
        english_text = item["translation"]["en"]
        
        # Only process non-empty texts
        if english_text and len(english_text.strip()) > 50:  # Filter very short texts
            doc = Document(
                page_content=english_text,
                metadata={
                    "doc_id": i,
                    "translation_pair": "nl_en_2016",
                    "length": len(english_text)
                }
            )
            documents.append(doc)

print(f"Created {len(documents)} documents from dataset")
print(f"Sample document: {documents[0].page_content[:200]}...")
```
